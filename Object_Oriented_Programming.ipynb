## **Constructor**

1. All classes have a function called __init__() , which is always executed when the object is being initiated.

2. Whenever we apply the concept of object , we automatically call the constructor.

3. Even if we apply manually apply the __init__ constructor itself we use the term self which is nothing but the object itself.  

4. The self parameter is a reference to the current instance of the class, and is used to access variables that belongs to the class.

5. The data that is stored inside the class or inside the objects, is called
   attributes.

6. Default constructors are by default that means even if we do not apply it
   manually , python will itself create it.

```python
class Student:
    college_name = "IIIT KOTTAYAM"
    name = "anonymous" #class attr

    # parameterized constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        print("adding new student in Database")

# creating objects outside the class
s1 = Student("karan", 97)
print(s1.name, s1.marks)

s2 = Student("arjun", 88)
print(s2.name, s2.marks)

# class variable
print(Student.college_name)
print(Student.name)
```

## **Encapsulation**

1. Wrapping data and functions into a single unit(object)
2. Combination of a class which includes attributes and methods + using objects is what we call as encapsulation.

```python
# Practice concepts
# Create an account class with 2 attributes - balance & account no.
# Create methods for debit, credit and printing the balance



class Account:

  def __init__(self , balance , acc):
    self.balance = balance
    self.account_no = acc

  def debit(self, amount):
    self.balance =- amount
    print("Rs: " , amount , "was debited" , self.get_balance())

  def credit(self , amount):
    self.balance =+ amount
    print("Rs. " , amount , "was credited" , self.get_balance())


  def get_balance(self):
    return self.balance


acc1 = Account(10000 , 12345)
print(acc1.balance)
print(acc1.account_no)

```

## **Private(like) attributes & methods**

**Conceptual Implementations in Python**


Private attributes & methods are meant to be used only within the class and are not accesible from outside the class

```python
class student:
  def __init__(self , acc_no , acc_pass):
    self.acc_no = acc_no
    self.__acc_pass = acc_pass   #private access cannot be accessed by objects

  def reset_pass(self):
    print(self.__acc_pass)

acc1 = student("12345" , "345")

print(acc1.acc_no)
print(acc1.reset_pass())
#print(acc1.__acc_pass)
```

## **Inheritance**

1. When one class derives the properties & methods of another class(parentheses).

2. Since we inherit the properties of class 1 to class 2 , even if we create object for class 2 and want to extract some methods from class 1 , we can easily do it.

```python
class Car:

  color = "black"

  @staticmethod
  def start():
     print("car started.. ")

  @staticmethod
  def stop():
    print("car stopeed..")


class Toyota_Car(Car):
  def __init__(self , name):
    self.name = name


car1 = Toyota_Car("fortuner")
car2 = Toyota_Car("prius")


print(car1.start())

print(car1.color)


```

## **Super Method**

1. super() method is used to access methods of the parent class

2. It is a special function which is important for inheriting the parent methods from different class

```python
class Car:

  def __init__(self, type):
      self.type = type

  @staticmethod
  def start():
    print("car started..")

  @staticmethod
  def stop():
    print("car stopped..")


class ToyotaCar(Car):
  def __init__(self, name, type):
    self.name = name
    super().__init__(type)


car1 = ToyotaCar("prius" , "electric")
print(car1.type)
```

## **dunder functions**

1. a+b (addition)
2. a-b (subtraction)
3. a*b (multiplication)
4. a/b (division)
5. a%b (addition)

```python
class Complex:
   def __init__(self, real, img):
      self.real = real
      self.img = img

   def showNumber(self):
      print(self.real, "i+", self.img, "j")

   def __add__(self, num2): #dunder functions
      newReal = self.real + num2.real
      newImg = self.img + num2.img
      return Complex(newReal, newImg)

   def __sub__(self, num2): #dunder functions
      newReal = self.real - num2.real
      newImg = self.img - num2.img
      return Complex(newReal, newImg)

num1 = Complex(1, 3)
num1.showNumber()

num2 = Complex(4, 6)
num2.showNumber()

num3 = num1+num2
num3.showNumber()

num4 = num1-num2
num4.showNumber()
```
